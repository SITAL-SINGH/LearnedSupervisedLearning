```python
import pandas as pd
from sklearn.model_selection import LeaveOneOut, LeavePOut, KFold, StratifiedKFold, cross_val_score
import seaborn as sns
from sklearn.preprocessing import StandardScaler
```

```python
data= pd.read_csv("../placement.csv")
```

```python
data.shape
```

```python
data.head()
```

```python
X=data.iloc[:,:-1]
y=data["package"]
```

```python
new_data = data.head(10)
```

```python
X=new_data.iloc[:,:-1]
y=new_data["package"]
```

```python
lo = LeaveOneOut ()
for train, test in lo.split(X,y):
    print(train, test)
```

```python
#leave p out
lp= LeavePOut(p=2)
for train, test in lp.split(X,y):
    print(train, test)
```

```python
#k_fold
kp = KFold(n_splits=5)
for train, test in kp.split(X,y):
    print(train, test)
```

```python
# to find the cross validation score
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LinearRegression
cross_val_score(LinearRegression(), X,y, cv=5)
```

```python
# kfold cross validation
cross_val_score(LinearRegression(), X,y, cv=KFold())
```
