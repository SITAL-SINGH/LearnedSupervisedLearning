```python
X = data.iloc[:,:-1]
y =data["Salary"]
X, y
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train, y_train)
```

```python
#out algo formde = y = m1.x1 + m2.x2 + c
m = lr.coef_
c = lr.intercept_
m,c
```

```python
lr.score(X_test, y_test)
```

```python
lr.predict(X_test)
```
