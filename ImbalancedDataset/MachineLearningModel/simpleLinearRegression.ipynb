```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

```

```python
data= pd.read_csv("../placement.csv")
```

```python
X = data[["cgpa"]]
y= data[["package"]]
X.shape, y.shape
```

```python
X.ndim, y.ndim
```

```python
sns.scatterplot(x="cgpa", y="package", data=data)
plt.show()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y , test_size=0.2, random_state=2)
X_train, X_test
```

```python
lr = LinearRegression()
```

```python
lr.fit(X_train, y_train)
```

```python
lr.predict([[6.89]])
```

```python
#accuracy score
lr.score(X_test, y_test)
```

```python
# to find the m and c in y = mx +c
m = lr.coef_
c = lr.intercept_
m, c
```

```python
 # making the scatter plot
sns.scatterplot(x="cgpa", y="package",data=data, label="original data")
plt.plot(data["cgpa"], lr.predict(X), label="predict line", color="red")
plt.show()
```
