```python
from sklearn.linear_model import LogisticRegression
```

```python
#making the object
model = LogisticRegression()

#sample input
X = [[1],[2],[3],[4],[5]]  #hours studied
y=[0,0,0,1,1] #pass or fail result

model.fit(X, y)
hrs = float(input("Enter the hours you studied"))
result = model.predict([[hrs]])[0]
if result==0:
    print(f"Studied hours: {hrs} and result: Fail")

else:
    print(f"Studied hours: {hrs} and result: Pass")
```
