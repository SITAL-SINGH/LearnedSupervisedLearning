```python
#program
model = LinearRegression()
X= [[1],[2],[3],[4],[5]]
y=[40,50,65,75,90]

#now training the model
model.fit(X, y)

studied_hours= float(input("Enter the hours you studied: "))

predict_marks = model.predict([[studied_hours]])

print(f"Studied hours: {studied_hours} and predicted marks: {predict_marks}")
```
