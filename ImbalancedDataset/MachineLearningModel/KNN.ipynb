```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
# weight and size of the fruit
X = [[180,7],[200, 75], [250 , 8], [300, 8.5], [330, 9], [360 , 95]]
# 0 = Apple , 1= orange

y= [0,0,0,1,1,1]

model = KNeighborsClassifier(n_neighbors=3)
model.fit(X, y)
weight = float(input("Enter the weight of the friut in gram: "))
size = float(input("Enter the size of the fruit in cm: "))
result = model.predict([[weight, size]])[0]
if result == 0:
    print("It is likely to be Apple")

else:
    print("It is likely to be orange")
```
