```python
import pandas as pd
```

```python
data = {
    'Name':['Ram','Sital','Shyam','Gyan','Man'],
    'Age':[23,None, 54, 23, None],
    'Salary': [1344, 2000, 3000, None, None]
}
```

```python
# creating the dataframe
df = pd.DataFrame(data)
print("original dataframe")
df
```

```python
# to find the total number of the missing values in the each column
df.isnull().sum()
```

```python
# to delete the missing values column
del_df = df.dropna()
del_df
```

```python
# no filling the value
df['Age'].fillna(df["Age"].mean(), inplace=True)
df['Salary'].fillna(df["Salary"].mean() , inplace=True)
df
```
