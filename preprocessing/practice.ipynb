```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import MinMaxScaler
from sklearn.preprocessing import LabelEncoder
```

```python
dataset = pd.read_csv("./sample_dataset.csv")
```

```python
dataset.head()
```

```python
dataset.isnull().sum()
```

```python
df_label = dataset.copy()


# label encoding of the Gender and passed column
le = LabelEncoder()
df_label["Gender_Encoded"] = le.fit_transform(df_label["Gender"])
df_label["Passed_Encoded"] = le.fit_transform(df_label["Passed"])
df_label.drop(["Name","Gender","City","Passed"], inplace=True, axis=1)
df_label
```

```python

sd_scaler = StandardScaler()
df_standard_scaled = sd_scaler.fit_transform(df_label)
pd.DataFrame(df_standard_scaled, columns=["Gender_Encoded", "Passed_Encoded"])
```

```python
minmax_scaler = MinMaxScaler()
minmax_scaled = minmax_scaler.fit_transform(df_label)
pd.DataFrame(minmax_scaled, columns=["Gender_Encoded", "Passed_Encoded"])
```
