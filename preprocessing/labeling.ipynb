```python
from sklearn.preprocessing import LabelEncoder
import pandas as pd
```

```python
df = pd.read_csv("./sample_dataset.csv")
```

```python
# making the copy of the original data
df_label = df.copy()
```

```python
le = LabelEncoder()
df_label["Gender_Encoded"] = le.fit_transform(df_label["Gender"])
df_label["Passed_Encoded"] = le.fit_transform(df["Passed"])
```

```python
print("Labeled Encoded Data: ")
print(df_label[["Name", "Gender", "Gender_Encoded", "Passed" , "Passed_Encoded"]].head())
```

```python
# one hot encoding
df_encoded = pd.get_dummies(df_label, columns=["City"])
print('One hot Encoding (City): ')
df_encoded.head()
```
