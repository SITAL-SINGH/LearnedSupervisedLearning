```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from mlxtend.plotting import plot_decision_regions
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB, BernoulliNB, MultinomialNB
```

```python
data = pd.read_csv(f"C:/Users/Sital/OneDrive/Desktop/Scikit-learn-fullcourse/placement (2).csv")
```

```python
data.head()
```

```python
data.drop(columns="Unnamed: 0", axis=1, inplace=True)
```

```python
plt.Figure(figsize=(10,5))
sns.scatterplot(x="cgpa", y="iq", hue="placement", data=data)
```

```python
X = data.iloc[:,:-1]
y=data["placement"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=2)
```

```python
# make the normal distribution plot
sns.kdeplot(data=data["cgpa"])
```

```python
sns.kdeplot(data=data["iq"])
```

```python
# Gaussian
gnb = GaussianNB()
gnb.fit(X_train, y_train)
```

```python
gnb.score(X_test, y_test)
```

```python
# Multinomial
mnb = MultinomialNB()
mnb.fit(X_train, y_train)
```

```python
mnb.score(X_test, y_test)
```

```python
# bernoulli
bnb = BernoulliNB()
bnb.fit(X_train, y_train)
```

```python
bnb.score(X_test, y_test)
```

```python
# decision region for gaussion
plot_decision_regions(X.to_numpy(), y.to_numpy(), clf=gnb)
```

```python
# decision for multinomial
plot_decision_regions(X.to_numpy(), y.to_numpy(), clf=mnb)
```

```python
# decision region for bernoulli
plot_decision_regions(X.to_numpy(), y.to_numpy(), clf=bnb)
```
