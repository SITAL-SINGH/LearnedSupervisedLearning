```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.svm import SVR
from sklearn.preprocessing import StandardScaler
from mlxtend.plotting import plot_decision_regions
```

```python
data = pd.read_csv(r"C:\Users\Sital\OneDrive\Desktop\Scikit-learn-fullcourse\placement.csv")
```

```python
data.head()
```

```python
sns.scatterplot(x="cgpa", y="package",data=data)
```

```python
X = data[["cgpa"]]
y=data["package"]
X
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2, random_state=2)
```

```python
svr = SVR(kernel="linear")
svr.fit(X_train, y_train)
```

```python
svr.score(X_test, y_test), svr.score(X_train, y_train)
```

```python
sns.scatterplot(x="cgpa", y="package",data=data)
plt.plot(data["cgpa"], svr.predict(X), color="red")
plt.show()
```

```python
svr1 = SVR(kernel="poly")
svr1.fit(X_train, y_train)
print(svr1.score(X_test, y_test), svr1.score(X_train, y_train))
sns.scatterplot(x="cgpa", y="package",data=data)
plt.plot(data["cgpa"], svr1.predict(X), color="red")
plt.show()
```

```python
#rbf
svr2 = SVR(kernel="rbf")
svr2.fit(X_train, y_train)
print(svr2.score(X_test, y_test), svr2.score(X_train, y_train))
sns.scatterplot(x="cgpa", y="package",data=data)
plt.plot(data["cgpa"], svr2.predict(X), color="red")
plt.show()
```

```python
#
svr4 = SVR(kernel="sigmoid")
svr4.fit(X_train, y_train)
print(svr4.score(X_test, y_test), svr4.score(X_train, y_train))
sns.scatterplot(x="cgpa", y="package",data=data)
plt.plot(data["cgpa"], svr4.predict(X), color="red")
plt.show()
```
