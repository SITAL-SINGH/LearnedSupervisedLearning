```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from mlxtend.plotting import plot_decision_regions
```

```python
data = pd.read_csv(r"C:\Users\Sital\OneDrive\Desktop\Scikit-learn-fullcourse\placement (2).csv")
```

```python
data.head()
```

```python
data.drop(columns="Unnamed: 0", axis=1, inplace=True)
```

```python
sns.scatterplot(x="cgpa", y='iq', data=data, hue="placement")
```

```python
X = data.iloc[:,:-1]
y= data["placement"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y , train_size=0.2, random_state=2)
```

```python
sv = SVC(kernel="linear")
sv.fit(X_train, y_train)
```

```python
sv.score(X_test, y_test),sv.score(X_train, y_train)
```

```python
plot_decision_regions(X.to_numpy(), y.to_numpy(), clf=sv)
```

```python
#polynomial
sv1 = SVC(kernel="poly")
sv1.fit(X_train, y_train)
print(sv1.score(X_test, y_test),sv1.score(X_train, y_train))


```

```python
plot_decision_regions(X.to_numpy(), y.to_numpy(), clf=sv1)
```

```python
# rbf
sv2 = SVC(kernel='rbf')
sv2.fit(X_train, y_train)
print(sv2.score(X_test, y_test),sv2.score(X_train, y_train))
```

```python
plot_decision_regions(X.to_numpy(), y.to_numpy(), clf=sv2)
```

```python
#sigmoid
sv3 = SVC(kernel='rbf')
sv3.fit(X_train, y_train)
print(sv3.score(X_test, y_test),sv3.score(X_train, y_train))
plot_decision_regions(X.to_numpy(), y.to_numpy(), clf=sv2)
```
